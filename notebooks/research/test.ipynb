```python
import torch

# 创建一个带梯度的张量（像是一个“会记笔记的x”）
x = torch.tensor(3.0, requires_grad=True)  

# 计算 y = x²（PyTorch会自动记录计算过程）
y = x ** 2  

# 自动计算梯度（即 dy/dx = 2x）
y.backward()  

# 查看x的梯度（应该等于 2*3=6）
print(x.grad)  # 输出：tensor(6.)
```

```python
import matplotlib.pyplot as plt

# x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
x = torch.arange(1, 3, 0.1, requires_grad=True)
y = x ** 2  

# 错误：绘图函数可能不接受带梯度的张量！
# plt.plot(x, y)  # 可能会报错  

# 正确：先 detach() 变成普通张量
plt.plot(x.detach(), y.detach())  # 正常绘图
```
